# The QA harness, one file at a time

**Seven notebooks, one per Python file in the harness. Every example runs.**

The engine set above this one (`../00-index.ipynb`) makes a specific claim: search `gl_engine/` for
`deductible` and you get one hit, because the engine executes ISO's rules rather than implementing
them. This set makes a different one -- **every defect found in the QA programme so far came from
something in these seven files.** The aggregate-limit ILF keying gap, OI-93's inert values, OI-91's
terrorism-place fallback: all in `variants.py` or `sweep.py`, not in the engine.

**Why a second set, rather than extending the first.** The engine notebooks are careful to keep no
rating concepts out of place -- the interpreter does not know what a deductible is. The harness is
the opposite kind of code: tiers, budgets, thinning, a run store. Mixing the two would blur a
distinction the project has been careful about. See `docs/UI-STRATEGY.md` for the fuller argument.

Start at [`01-variants`](01-variants.ipynb) and read down; each notebook may assume the ones above
it.

## Two things worth knowing before you run one

**Notebooks that would write to the real results store redirect it to a temporary directory first**
([`05`](05-runstore.ipynb), [`07`](07-layers.ipynb)) -- a demo run landing in `results/` every time
someone opens the notebook would corrupt the thing the store exists to keep honest.

**Outputs are stripped before commit, same as the engine set.** A notebook that has run holds ISO's
licensed premiums inside its JSON. See [`../README.md`](../README.md).

**They are tested.** `python tests/verify_notebooks.py` executes every cell of every notebook in
both sets and fails on any exception.
